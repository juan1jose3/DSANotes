#### Creation of a linked list:


#### Node class:

```python
class Node:
    def __init__(self,data, next=None):
        self.data = data
        self.next = next

    def get_next(self):
        return self.next
    
    def set_next(self, new_next):
        self.next = new_next

    
    def get_data(self):
        return self.data
    
    def set_data(self, new_data):
        self.data = new_data
```

#### Linked_list class:

```python
class Linked_list:
    def __init__(self,root=None):
        self.root = root
        self.size = 0
    
    def get_root(self):
        return self.root
    
    def traversal(self):
        current_node = self.root

        while current_node:
            print(current_node.get_data())
            current_node = current_node.get_next()


    # adding from beginning
    
    
    # find
    def find(self, data_to_find):
        current_node = self.root
        while current_node:
            if current_node.get_data() == data_to_find:
                return current_node.get_data()
            current_node = current_node.get_next()
        return "NOT FOUND"
    
    # challenges 

    def insert_end(self,data_to_insert):
        current_node = self.root

        while current_node:
            if current_node.get_next() == None:
                new_node = Node(data_to_insert,None)
                current_node.set_next(new_node)
                self.size += 1 
                return f"Node added {data_to_insert}"
            current_node = current_node.get_next()
    
```

#### Add to a linked list:

```python
def add(self,data):
        new_node = Node(data,self.root)
        self.root = new_node
        self.size += 1

Linked_list.add = add
```

```python
linked_list = Linked_list()
linked_list.add(50)
linked_list.add(40)
linked_list.add(30)
linked_list.add(20)
linked_list.add(11)

linked_list.traversal()
```

#### Remove from a linked list:

```python
# remove from linked list
def remove(self,data):
    current_node = self.root
    prev_node = None

    while current_node:
        if current_node.get_data() == data:
            if prev_node:
                prev_node.set_next(current_node.next)
                self.size -= 1
            else:
                self.root = current_node.get_next()
                self.size -= 1
            return True
        else:
            prev_node = current_node
            current_node = current_node.get_next()
    return False

Linked_list.remove = remove
```

```python
linked_list = Linked_list()
linked_list.add(22)
linked_list.add(43)
linked_list.add(3)
linked_list.add(2)
linked_list.add(15)

linked_list.remove(43)

linked_list.traversal()
```

#### Find middle:

```python
def find_middle(self):
    fast = self.root
    slow = self.root

    while fast and fast.get_next():
        fast = fast.get_next().get_next()
        slow = slow.get_next()

    return slow.get_data()

Linked_list.find_middle = find_middle


```

```python
linked_list = Linked_list()
linked_list.add(50)
linked_list.add(40)
linked_list.add(30)
linked_list.add(20)
linked_list.add(10)

print(linked_list.find_middle())
```

#### Find last node

```python
def find_last_node(self):
    current_node = self.root
    while current_node:
        if current_node.get_next() == None:
            return current_node.get_data()
        current_node = current_node.get_next()

Linked_list.find_last_node = find_last_node
```

```python
linked_list = Linked_list()
linked_list.add(12)
linked_list.add(54)
linked_list.add(30)
linked_list.add(11)

print(linked_list.find_last_node())
```

#### Get size by walking

```python
def get_size_by_walking(self):
    hops = 0
    current_node = self.root
    
    while current_node:
        hops += 1
        current_node = current_node.get_next()
    return hops

Linked_list.get_size_by_walking = get_size_by_walking
```

```python
linked_list = Linked_list()
linked_list.add(12)
linked_list.add(54)
linked_list.add(30)
linked_list.add(11)
linked_list.add(436)

print(linked_list.get_size_by_walking())
```
